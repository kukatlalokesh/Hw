PCA requires standardization when features have different scales. Cluster labels are arbitrary.

```python
from statistics import mean, pstdev
values = [10, 12, 11, 13, 12, 100]
avg, std = mean(values), pstdev(values)
print([(v, round((v-avg)/std, 2)) for v in values])
```
