# Week 2 NumPy and Pandas Summary

This notebook documents the NumPy and pandas concepts practiced during Week 2.

```python
import numpy as np
import pandas as pd

array = np.array([[10, 20, 30], [40, 50, 60]])
print('Shape:', array.shape)
print('Size:', array.size)
print('Dimensions:', array.ndim)
print('Reshaped:', array.reshape(3, 2))
print('Flattened:', array.flatten())
```

```python
df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Score': [88, 76, 95],
    'Department': ['IT', 'HR', 'IT']
})

print(df.head())
print(df.describe())
print(df.groupby('Department')['Score'].mean())
```
